```python
def read_s3_parquet(aws_access_key_id, aws_secret_access_key, path):
    '''
    
    
    
    '''
    source_path = (
        f"s3a://{aws_access_key_id}:{aws_secret_access_key}"
        f"@{path}"
    )
    df = spark.read.parquet(source_path)
    return df
```
